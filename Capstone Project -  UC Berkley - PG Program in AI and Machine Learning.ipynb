###Capstone Assignment: Customer Segmentation for E-commerce

**Problem Statement - The research question**

In today’s fast-moving e-commerce world, many businesses still use the same marketing approach for all customers. This often leads to poor engagement and low sales because it doesn’t reflect how different people shop. To fix this, goal is to use a customer shopping dataset, which includes useful details like age, gender, location, how much people spend, how often they shop, what they buy, what category of items, sizes etc, and how they pay. The goal is to build a smart system that groups customers based on their shopping habits. These groups will help businesses send more personalized offers that match each customer’s style and needs.
Business Goal

The main goal of this project is to build a machine learning model that groups customers based on how they shop, how often they buy, how much they spend, and how recently they made a purchase (known as RFM: Recency, Frequency, Monetary).
Why this question is important?

By creating clear customer segments, the marketing team can stop using generic messages and start sending personalized offers that match each group’s habits. This approach is expected to boost repeat purchases and increase overall sales by at least 10%.

### Calculate IQR

### Calculate the Interquartile Range (IQR) for the numeric columns.


**Reasoning**:
Calculate Q1, Q3, and IQR for the numeric columns to identify outliers.



```python
# Select numeric columns
numeric_cols = Data.select_dtypes(include=['number'])

# Calculate Q1 (25th percentile)
Q1 = numeric_cols.quantile(0.25)

# Calculate Q3 (75th percentile)
Q3 = numeric_cols.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

print("First Quartile (Q1):")
display(Q1)

print("\nThird Quartile (Q3):")
display(Q3)

print("\nInterquartile Range (IQR):")
display(IQR)
```

### Identify outliers using IQR

#### Define a range based on the IQR (e.g., 1.5 * IQR below the first quartile and 1.5 * IQR above the third quartile) and identify data points outside this range as outliers.


**Reasoning**:
Calculate the lower and upper bounds for outlier detection using the IQR method and identify outliers.



```python
# Calculate the lower bound for outlier detection (Q1 - 1.5 * IQR)
lower_bound = Q1 - 1.5 * IQR

# Calculate the upper bound for outlier detection (Q3 + 1.5 * IQR)
upper_bound = Q3 + 1.5 * IQR

print("Lower Bounds for Outlier Detection (IQR Method):")
display(lower_bound)

print("\nUpper Bounds for Outlier Detection (IQR Method):")
display(upper_bound)

# Initialize an empty DataFrame to store identified outliers
iqr_outliers = pd.DataFrame()

# Iterate through each numeric column and identify outliers
for col in numeric_cols.columns:
    # Identify rows where the column value is less than the lower bound or greater than the upper bound
    outlier_rows = Data[(Data[col] < lower_bound[col]) | (Data[col] > upper_bound[col])].copy()

    # Add a column to indicate the column in which the outlier was detected
    outlier_rows['Outlier Column'] = col

    # Concatenate these outlier rows to the overall outlier DataFrame
    iqr_outliers = pd.concat([iqr_outliers, outlier_rows], ignore_index=True)

print("\nIdentified Outliers (IQR Method):")
display(iqr_outliers)
```

### Calculate Z-Scores

### Calculate the Z-score for each data point in the numeric columns.


**Reasoning**:
Calculate the Z-score for each data point in the numeric columns using the zscore function from scipy.stats and store the result in a new DataFrame.



```python
from scipy.stats import zscore

# Calculate Z-scores for each element in numeric_cols
z_scores = numeric_cols.apply(zscore)

# Display the first few rows of the z_scores DataFrame
print("First few rows of Z-scores DataFrame:")
display(z_scores.head())
```

### Identify outliers using z-scores

### Define a threshold for the absolute Z-score and identify data points with absolute Z-scores above this threshold as outliers.


**Reasoning**:
Define a Z-score threshold, iterate through numeric columns, identify outliers based on the threshold, retrieve original rows, add an 'Outlier Column', concatenate to an outlier DataFrame, and display the results.



```python
# Define a Z-score threshold (e.g., 3)
z_score_threshold = 3

# Initialize an empty DataFrame to store identified outliers
zscore_outliers = pd.DataFrame()

# Iterate through each numeric column for which Z-scores were calculated
for col in z_scores.columns:
    # Identify the rows where the absolute Z-score is greater than the defined threshold
    outlier_indices = z_scores[abs(z_scores[col]) > z_score_threshold].index

    # Retrieve the original data rows from the Data DataFrame corresponding to the identified outlier indices
    outlier_rows = Data.loc[outlier_indices].copy()

    # Add a column to the outlier rows DataFrame to indicate the column in which the outlier was detected
    outlier_rows['Outlier Column'] = col

    # Concatenate these outlier rows to the overall outlier DataFrame
    zscore_outliers = pd.concat([zscore_outliers, outlier_rows], ignore_index=True)

# Display the identified outliers (the populated outlier DataFrame)
print(f"\nIdentified Outliers (Z-score > {z_score_threshold}):")
display(zscore_outliers)
```

**Identified Outliers (Z-score Method):**

The Z-score method, using a threshold of 3, did not identify any outliers in the dataset. The `zscore_outliers` DataFrame is empty.

**Comparison and Discussion:**

The significant difference in the number of outliers detected by the two methods highlights their distinct approaches.

*   **IQR Method:** This method is based on the distribution of the data and is less sensitive to extreme values when calculating the range (as it uses quartiles). It's good at finding values that are far from the bulk of the data. It identified the intentionally added outliers in 'Age' and 'Purchase Amount (USD)'.
*   **Z-score Method:** This method measures how many standard deviations away from the mean a data point is. It is highly sensitive to extreme values because the mean and standard deviation are heavily influenced by outliers. In this case, it appears that even the intentionally added extreme values in 'Age' and 'Purchase Amount (USD)', while identified as outliers by the IQR method, were not extreme enough relative to the mean and standard deviation of the *entire* augmented dataset (including the original data and the new rows) to exceed a Z-score threshold of 3. The standard deviation might be large enough due to the spread of the augmented data to keep the Z-scores of these points below the threshold.

**Potential Reasons for Discrepancies:**

1.  **Sensitivity to Extreme Values:** The Z-score is more sensitive to extreme values which can inflate the standard deviation and make it harder for individual data points to exceed the threshold. The IQR, based on quartiles, is more robust to extreme values.
2.  **Data Distribution:** The distribution of the augmented data, while containing some extreme values, might not have a shape where those extremes translate to Z-scores greater than 3 across the board.

In conclusion, the IQR method was more effective in identifying the intentionally introduced outliers in 'Age' and 'Purchase Amount (USD)' in this specific augmented dataset. The Z-score method, with a threshold of 3, did not flag any data points as outliers. This demonstrates that the choice of outlier detection method and its parameters (like the Z-score threshold) can significantly impact the results, and the most appropriate method depends on the data distribution and the specific goals of the analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA # For 2D visualization
from mpl_toolkits.mplot3d import Axes3D # For 3D plotting
import warnings
import os

# --- 1. Select Input Features ---

#Features you can use choose from below for KMeans
#input_features = ['Age','Gender','Item Purchased', 'Category', 'Purchase Amount (USD)', 'Location', 'Size',
#                    'Color', 'Season', 'Review', 'Rating', 'Subscription Status', 'Payment Method',
#                    'Shipping Type', 'Discount Applied', 'Previous Purchases', 'Preferred Payment Method',
#                    'Frequency of Purchases', 'Purchase Recency', 'Avg Spend Per Purchase']

#input_features = ['Item Purchased', 'Size', 'Avg Spend Per Purchase']
#input_features = ['Gender', 'Discount Applied']
#input_features = ['Purchase Amount (USD)', 'Location']
#input_features = ['Purchase Amount (USD)', 'Location', 'Payment Method']

#Below most co-related features selected as seen from the heap-map above
input_features = [ 'Location',
                     'Season', 'Payment Method',
                    'Shipping Type', 'Preferred Payment Method',
                    'Avg Spend Per Purchase', 'Purchase Amount (USD)']

# --- 2. Select and Scale Features ---
# Select only the columns from 'Data' that you want to use for clustering
X_data = Data[input_features]

# K-Means is a distance-based algorithm.
# It's CRITICAL to scale your features so they have equal weight.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_data)

# Convert scaled data back to a DataFrame for clarity (optional, but good practice)
X_scaled_df = pd.DataFrame(X_scaled, columns=input_features)

print(f"--- Scaled Data (using {', '.join(input_features)}) ---")
display(X_scaled_df.head())
print("\n")

```

#### Evaluate K and Elbow Curve

```python

# --- 3. Find the Optimal Number of Clusters (k) using the Elbow Method ---
# We'll test k from 1 to 10
inertia_values = []
possible_k_values = range(1, 11)

for k in possible_k_values:
    # n_init=10 is the default and recommended to avoid bad initializations
    kmeans = KMeans(n_clusters=k, init='k-means++', n_init='auto', random_state=42)
    kmeans.fit(X_scaled_df)
    # inertia_ is the "Within-Cluster Sum of Squares" (WCSS)
    inertia_values.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(10, 6))
plt.plot(possible_k_values, inertia_values, 'bo-') # 'bo-' creates blue dots and a line
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia (WCSS)')
plt.xticks(possible_k_values)
plt.grid(True)
plt.show()
plt.close()
```

#### Run KMeans with Optimal K

```python
# --- 4. Run K-Means with the Optimal k ---

# Based on the elbow plot, choose k
OPTIMAL_K = 3

# Initialize and fit the final model
kmeans_final = KMeans(n_clusters=OPTIMAL_K, init='k-means++', n_init='auto', random_state=42)
kmeans_final.fit(X_scaled_df)

# Get the cluster assignments for each data point
cluster_labels = kmeans_final.labels_

# --- 5. Add Cluster Labels back to the Original DataFrame ---
# This is the final result, linking clusters to your original data
Data['KMeans-Cluster'] = cluster_labels
display(Data_Cleaned.head())
#Save the KMeans Cluster in the original cleaned data
Data_Cleaned ['KMeans-Cluster'] = cluster_labels
#Data_Cleaned.head()
```

```python
# --- 6. Visualize the Clusters ---

# Create figure and title
fig = plt.figure(figsize=(12, 8))
title = f'K-Means Clustering (k={OPTIMAL_K}) on {len(input_features)} Features'
palette = 'viridis' # Use a color-blind friendly palette

if len(input_features) == 1:
    # Case for 1 feature
    print("--- Visualizing 1D clusters ---")
    ax = fig.add_subplot(111)
    sns.stripplot(
        data=Data,
        x=input_features[0],
        y='KMeans-Cluster',
        hue='KMeans-Cluster',
        palette=palette,
        jitter=0.2,
        s=10,
        legend=False,
        ax=ax
    )
    ax.set_title(title)
    ax.grid(True)

elif len(input_features) == 2:
    # Easiest case: We can plot the 2 features directly
    print("--- Visualizing 2D clusters ---")
    ax = fig.add_subplot(111)
    sns.scatterplot(
        data=Data,
        x=input_features[0],
        y=input_features[1],
        hue='KMeans-Cluster',
        palette=palette,
        s=100,
        alpha=0.7,
        legend='full',
        ax=ax
    )
    # Plot the centroids
    centers = scaler.inverse_transform(kmeans_final.cluster_centers_) # Un-scale centers
    ax.scatter(
        centers[:, 0],
        centers[:, 1],
        c='black',
        s=250,
        marker='X',
        label='Centroids'
    )
    ax.set_title(title)
    ax.legend()
    ax.grid(True)

elif len(input_features) == 3:
    # --- NEW 3D PLOT LOGIC ---
    print("--- Visualizing 3D clusters ---")
    # Need to clear the 2D figure and create a 3D one
    plt.close(fig) # Close the 2D fig
    fig = plt.figure(figsize=(12, 9)) # Make a new one

    ax = fig.add_subplot(111, projection='3d')

    # Get unique cluster labels and assign colors
    clusters = sorted(Data['KMeans-Cluster'].unique())
    # Create a color map
    colors = plt.cm.get_cmap(palette, len(clusters))
    color_map = {cluster: colors(i) for i, cluster in enumerate(clusters)}

    # Plot data points
    ax.scatter(
        Data[input_features[0]],
        Data[input_features[1]],
        Data[input_features[2]],
        c=Data['KMeans-Cluster'].map(color_map),
        s=50,
        alpha=0.6
    )

    # Plot the centroids
    centers = scaler.inverse_transform(kmeans_final.cluster_centers_) # Un-scale centers
    ax.scatter(
        centers[:, 0],
        centers[:, 1],
        centers[:, 2],
        c='red',
        s=250,
        marker='X',
        label='Centroids',
        depthshade=True # Gives a bit of 3D effect to the marker
    )

    # Create proxy artists for legend (since scatter's 'c' was an array)
    legend_handles = [plt.Line2D([0], [0], marker='o', color='w',
                                 label=f'Cluster {c}',
                                 markerfacecolor=color_map[c], markersize=10)
                      for c in clusters]
    legend_handles.append(plt.Line2D([0], [0], marker='X', color='w',
                                     label='Centroids',
                                     markerfacecolor='red', markersize=12))

    ax.set_title(title)
    ax.set_xlabel(input_features[0])
    ax.set_ylabel(input_features[1])
    ax.set_zlabel(input_features[2])
    ax.legend(handles=legend_handles)
    # 3D plot has grid by default

elif len(input_features) > 3:
    # --- MODIFIED: Was > 2, now > 3 ---
    # We must reduce dimensions to 2D for plotting
    print(f"--- Using PCA to visualize {len(input_features)} dimensions in 2D ---")
    ax = fig.add_subplot(111)
    pca = PCA(n_components=2, random_state=42)
    X_pca = pca.fit_transform(X_scaled_df)

    # Create a new DataFrame for plotting
    pca_df = pd.DataFrame(X_pca, columns=['PCA1', 'PCA2'])
    pca_df['KMeans-Cluster'] = cluster_labels

    sns.scatterplot(
        data=pca_df,
        x='PCA1',
        y='PCA2',
        hue='KMeans-Cluster',
        palette=palette,
        s=100,
        alpha=0.7,
        legend='full',
        ax=ax
    )

    # Add cluster centers (transformed to PCA space)
    centers_pca = pca.transform(kmeans_final.cluster_centers_)
    ax.scatter(
        centers_pca[:, 0],
        centers_pca[:, 1],
        c='black',
        s=250,
        marker='X',
        label='Centroids'
    )
    ax.set_title(title + ' (Visualized with PCA)')
    ax.legend()
    ax.grid(True)

# Save the plot
plt.show()
plt.close()
```

#### Evaulation Metrics

```python

from sklearn.metrics import silhouette_score, davies_bouldin_score

print("\n--- 7. Evaluation Metrics - K-Means Clustering ---")

# --- Silhouette Score ---
# Range: -1 to +1
# +1: Best. Clusters are dense and well-separated.
#  0: Neutral. Clusters overlap.
# -1: Worst. Data points may be assigned to the wrong clusters.
# Goal: Maximize.
silhouette_avg = silhouette_score(X_scaled_df, cluster_labels)
print(f"Silhouette Score: {silhouette_avg:.4f}")

# --- Davies-Bouldin Index ---
# Range: 0 to +Infinity
#  0: Best.
# Measures the average 'similarity' between clusters, where similarity
# is a ratio of within-cluster distances to between-cluster distances.
# Goal: Minimize.
db_index = davies_bouldin_score(X_scaled_df, cluster_labels)
print(f"Davies-Bouldin Index: {db_index:.4f}")
```

####Prepare the input features

```python
# Import hierarchical clustering modules
from sklearn.cluster import AgglomerativeClustering
import scipy.cluster.hierarchy as sch

# --- 1. Select Input Features ---

#Features you can use choose from below for KMeans
#input_features = ['Age','Gender','Item Purchased', 'Category', 'Purchase Amount (USD)', 'Location', 'Size',
#                    'Color', 'Season', 'Review Rating', 'Subscription Status', 'Payment Method',
#                    'Shipping Type', 'Discount Applied', 'Previous Purchases', 'Preferred Payment Method',
#                    'Frequency of Purchases', 'Purchase Recency', 'Avg Spend Per Purchase']

#input_features = ['Item Purchased', 'Size', 'Avg Spend Per Purchase']
#input_features = ['Gender', 'Discount Applied']
#input_features = ['Purchase Amount (USD)', 'Location']
#input_features = ['Purchase Amount (USD)', 'Location', 'Payment Method']

#Below most co-related features as seen from the heap-map above
input_features = [ 'Location',
                     'Season', 'Payment Method',
                    'Shipping Type', 'Preferred Payment Method',
                    'Avg Spend Per Purchase', 'Purchase Amount (USD)']

#input_features = ['Age', 'Season', 'Purchase Amount (USD)']


# --- 2. Select and Scale Features ---
# Select only the columns from 'Data' that you want to use for clustering
X_data = Data[input_features]

# Scaling is also critical for hierarchical clustering
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_data)

# Convert scaled data back to a DataFrame for clarity
X_scaled_df = pd.DataFrame(X_scaled, columns=input_features)

print("--- Hierarchical Clustering : Selected Features scaled Data ---")
display(X_scaled_df.head())
```

#### Generate the Dendrogram to find Optimal K

```python
# --- 3. Generate the Dendrogram to Find Optimal 'k' ---
# A dendrogram is the primary tool for hierarchical clustering
print("--- Hierarchical Clustering : Generating Dendrogram ---")
plt.figure(figsize=(15, 7))
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points (Index)")
plt.ylabel("Euclidean Distance (Ward's Linkage)")

# 'ward' linkage minimizes the variance of the clusters being merged
# This is similar in concept to K-Means' inertia
dendrogram = sch.dendrogram(sch.linkage(X_scaled_df, method='ward'))

# Draw a line to suggest a cut-off
# We'll look for the longest vertical line that doesn't cross other clusters
# Here, a cut at ~50 would give 3 clear clusters
plt.axhline(y=50, color='r', linestyle='--', label='Cut-off for k')
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.7)
plt.show()
plt.close()
```

#### Run Hierarchical Clustering with Optimal K

```python
# --- 4. Run Hierarchical Clustering with the Optimal k ---

# *** THIS VALUE TO BE CHANGED BASED ON  DENDROGRAM ***
OPTIMAL_K = 3

# Initialize and fit the final model

hierarchical_model = AgglomerativeClustering(n_clusters=OPTIMAL_K, linkage='ward')

# fit_predict gives us the cluster labels
# This model does NOT have a 'cluster_centers_' attribute like K-Means
cluster_labels = hierarchical_model.fit_predict(X_scaled_df)


# --- 5. Add Cluster Labels back to the Original DataFrame ---
# This is the final result, linking clusters to your original data
Data['Hierarchical-Cluster'] = cluster_labels
#Save Hierarchical Cluster columns in Original Data
Data_Cleaned['Hierarchical-Cluster'] = cluster_labels
```
